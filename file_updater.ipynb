```python
import os
from datetime import datetime
import shutil
from pathlib import Path
from dotenv import dotenv_values

today_str = datetime.now().strftime("%Y%m%d")
config = dotenv_values(".env")

BASE_DIR = Path.cwd() / "shutterstock"
BATCH_ID = f"batch_{today_str}"

windows_source = f'{config['SOURCE_DIR']}'
wsl_destination = f'{BASE_DIR}/input/{BATCH_ID}'

if not os.path.exists(windows_source):
    raise FileNotFoundError(f"Source directory does not exist: {windows_source}")
else:
    source_status = "OK!!"

if not os.path.exists(wsl_destination):
    os.makedirs(wsl_destination, exist_ok=True)
    destination_status = "Created!!"
else:
    destination_status = "OK!!"

print("source_status:", source_status)
print("destination_status:", destination_status)
```

```python
def copy_windows_to_wsl(source, destination):
    try:
        shutil.copytree(source, destination, dirs_exist_ok=True)
        print(f"✓ Successfully copied all files from {source} to {destination}")
        
        file_count = sum([len(files) for _, _, files in os.walk(destination)])
        print(f"Total files in destination: {file_count}")
        
    except Exception as e:
        print(f"✗ Error copying files: {str(e)}")
        raise

# Execute the copy
copy_windows_to_wsl(windows_source, wsl_destination)
```

```python
INPUT_DIR = BASE_DIR / "input" / BATCH_ID
OUTPUT_DIR = BASE_DIR / "output" / BATCH_ID
AI_DIR = BASE_DIR / "ai_output" / BATCH_ID

INPUT_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
AI_DIR.mkdir(parents=True, exist_ok=True)

print("Input:", INPUT_DIR)
print("Output:", OUTPUT_DIR)
```

```python
try:    
    with open(".env", "a") as f:
        f.write(f'\nBATCH_ID="{BATCH_ID}"')
        f.write(f'\nBASE_DIR="{BASE_DIR}"')
        f.write(f'\nINPUT_DIR="{INPUT_DIR}"')
        f.write(f'\nOUTPUT_DIR="{OUTPUT_DIR}"')
except Exception as e:
    print(str(e))
    raise
```
